Each context is annotated with multiple question–answer pairs following the SQuAD v2.0 format, including both answerable and unanswerable questions to improve model robustness and generalisation.

Finally, the dataset was saved in a structured JSON format following the SQuAD schema, ensuring compatibility with the model training and evaluation pipeline.

 **Data Augmentation (Question Expansion)**

After manually generating the QA pairs,  we had only **2,889 QA pairs**, which does not meet the **CourseWork2** required dataset size (3000-8000). To address this limitation, a fast and scalable data augmentation approach to increase the number of question–answer pairs is adopted .

The approach focuses on **question paraphrasing** by prepending different explanatory prefixes to existing answerable questions,which is randomly applying a set of prefixes such as *"Explain"*, *"Describe"*, and *"What is meant by"* to generate new samples.This preserves the original meaning while creating diverse variations of the questions. 

After applying this augmentation strategy, the dataset size increased significantly to a total of **4,727 QA pairs**, consisting of **3,676 answerable** questions and **1,051 unanswerable** questions.

```python
import copy
import random

# ✅ Better question variations
prefixes = [
    "Explain",
    "Describe",
    "What is meant by",
    "Give an explanation of",
    "How would you describe",
    "What is",
    "What causes",
    "What is the role of",
    "What defines"
]

new_data = []

for qa in full_dataset:
    if not qa["is_impossible"]:
        new_q = copy.deepcopy(qa)

        prefix = random.choice(prefixes)
        new_q["question"] = prefix + " " + qa["question"].lower()

        new_data.append(new_q)

# ✅ Add new questions to dataset
full_dataset.extend(new_data)

# ✅ Total count AFTER expansion
total = len(full_dataset)
answerable = sum(1 for x in full_dataset if not x["is_impossible"])
unanswerable = sum(1 for x in full_dataset if x["is_impossible"])

# ✅ Print results
print("✅ New total QA pairs:", total)
print("✅ Answerable:", answerable)
print("✅ Unanswerable:", unanswerable)
```

**Dataset Serialization (convert to JSON Format)**

After completing dataset expansion, the next step involved structuring the data into the **SQuAD (Stanford Question Answering Dataset) format**, which is widely used for training and evaluating QA models.

First, the dataset was **grouped by context**, to ensure  that all question–answer pairs associated with the same text passage were organized together. Each question was assigned a unique identifier and retained its attributes, including the question text, answers, and whether it is answerable.

Next, the grouped data was reformatted into the standard SQuAD structure, where each context is stored under a title and contains a list of corresponding QA pairs within a `paragraphs` field.

Finally, the structured dataset was saved as a JSON file named **`Pressure_Ulcer_dataset.json`**, making it ready for use in machine learning models and QA systems.

```python
import json
from collections import defaultdict 

# 1. ✅Group the questions by their specific context
grouped = defaultdict(list)

for i, qa in enumerate(full_dataset):
    grouped[qa["context"]].append({
        "id": f"q_{i}",
        "question": qa["question"],
        "answers": qa["answers"],
        "is_impossible": qa["is_impossible"]
    })

# ✅ Convert to SQuAD format
data = []

for i, (context, qas) in enumerate(grouped.items()):
    data.append({
        "title": f"context_{i}",
        "paragraphs": [
            {
                "context": context,
                "qas": qas
            }
        ]
    })

# 3. Create the final dictionary structure
dataset = {"data": data}

# ✅ Save file
with open("Pressure_Ulcer_dataset.json", "w") as f:
    json.dump(dataset, f, indent=4)

print("✅ Dataset saved in SQuAD format ✅")
```

## 3. Data Pre-processing


In this section, the dataset was prepared for model training through a series of preprocessing steps, including **converting the dataset into a flat structure** (where each entry contains a single context, question, and answer for easier handling), applying **tokenization** (breaking the text into smaller units that the model can process), and adding the **start and end positions** of the answers within the context (to help the model identify the exact location of each answer in the text).

**Convert SQuAD JSON  to Flat dataset**


In this step, the dataset was converted from the hierarchical SQuAD format into a flat structure using the Hugging Face `Dataset` library,to simplifies data handling and prepares the dataset for further preprocessing and model training.

Each entry now contains a single context, question, answer, and its corresponding label.

As shown in the cell below ,our conversion was successful, resulting in a dataset with **4,727 rows** and the following features:

- context  
- question  
- answers  
- is_impossible 



```python
#Convert SQuAD JSON  to Flat dataset
from datasets import Dataset

data_list = []

for item in dataset["data"]:
    for para in item["paragraphs"]:
        context = para["context"]
        for qa in para["qas"]:
            data_list.append({
                "context": context,
                "question": qa["question"],
                "answers": qa["answers"],
                "is_impossible": qa["is_impossible"]
            })

hf_dataset = Dataset.from_list(data_list)

print("✅ Dataset converted successfully!")

#Check
print(hf_dataset)

```

**TOKENIZATION**


In this step, tokenization was applied using the **BERT tokenizer (`bert-base-uncased`)** to convert the text into a format suitable for the model. Both the questions and contexts were transformed into tokens that can be processed numerically.

To handle long contexts, a **sliding window approach** was used, ensuring that no important information is lost when the context exceeds the maximum length. Additionally, **offset mappings** were included to keep track of the positions of words in the original text.


```python
# LOAD TOKENIZER
from transformers import BertTokenizerFast
# Using bert-base-uncased 
tokenizer = BertTokenizerFast.from_pretrained("bert-base-uncased")

# CREATE TOKENIZATION FUNCTION
def preprocess_function(examples):
    questions = [q.strip() for q in examples["question"]]
    
    # We use return_overflowing_tokens to handle long clinical contexts 
    # and return_offsets_mapping to link answer strings back to token IDs.
    tokenized_examples = tokenizer(
        questions,
        examples["context"],
        max_length=384,
        truncation="only_second",
        stride=128,
        return_overflowing_tokens=True,
        return_offsets_mapping=True,
        padding="max_length",
    )
    return tokenized_examples

# APPLY TOKENIZATION
#  remove original columns because the sliding window creates more rows than we started with
tokenized_dataset = hf_dataset.map(
    preprocess_function, 
    batched=True, 
    remove_columns=hf_dataset.column_names
)

print("✅ Tokenization complete!")
```

A quick check was performed to ensure that the data was properly tokenized to a fixed length of **384 tokens**, and that the required BERT special tokens **[CLS]** and **[SEP]** were correctly included.

The results  confirm that the tokenization step was applied successfully.

```python
# Select the first sample to verify
check_sample = tokenized_dataset[0]

# Decode the IDs back to text
decoded_text = tokenizer.decode(check_sample["input_ids"])

print("TOKENIZATION VERIFICATION ")
print(f"✅ Sample 0 Length: {len(check_sample['input_ids'])} tokens")
print(f"✅ Decoded Text Preview:\n{decoded_text[:250]}...")

# Technical confirmation
has_cls = tokenizer.cls_token_id in check_sample["input_ids"]
has_sep = tokenizer.sep_token_id in check_sample["input_ids"]
print(f"\n✅ Structure: CLS present: {has_cls} | SEP present: {has_sep}")
```

**Add stop & End positons**


In this step, the start and end positions of each answer were added to the tokenized dataset to indicate exactly where the answer appears within the context.

This was done by mapping the character positions of each answer in the original text to their corresponding token positions after tokenization,which is  important  in extractive question answering tasks because the model processes tokens rather than raw text, so it needs to know which tokens form the correct answer.

For unanswerable questions, both start and end positions were set to the **[CLS] token**, allowing the model to learn when no answer exists in the context.

After this step, the final processed dataset contains **4,727 samples** with the required features: `input_ids`, `token_type_ids`, `attention_mask`, `start_positions`, and `end_positions`, confirming that our data is fully prepared.


```python
# Function to align answer characters to token positions
def add_positions(tokenized_examples):
    sample_mapping = tokenized_examples.pop("overflow_to_sample_mapping")
    offset_mapping = tokenized_examples.pop("offset_mapping")

    tokenized_examples["start_positions"] = []
    tokenized_examples["end_positions"] = []

    for i, offsets in enumerate(offset_mapping):
        input_ids = tokenized_examples["input_ids"][i]
        cls_index = input_ids.index(tokenizer.cls_token_id)
        
        # Manually find [SEP] tokens to locate the clinical context
        sep_indices = [idx for idx, val in enumerate(input_ids) if val == tokenizer.sep_token_id]
        
        sample_index = sample_mapping[i]
        answers = hf_dataset[sample_index]["answers"]
        is_impossible = hf_dataset[sample_index]["is_impossible"]

        if is_impossible or len(answers) == 0:
            tokenized_examples["start_positions"].append(cls_index)
            tokenized_examples["end_positions"].append(cls_index)
        else:
            start_char = answers[0]["answer_start"]
            end_char = start_char + len(answers[0]["text"])

            token_start_index = sep_indices[0] + 1
            token_end_index = sep_indices[1] - 1

            if not (offsets[token_start_index][0] <= start_char and offsets[token_end_index][1] >= end_char):
                tokenized_examples["start_positions"].append(cls_index)
                tokenized_examples["end_positions"].append(cls_index)
            else:
                while token_start_index < len(offsets) and offsets[token_start_index][0] <= start_char:
                    token_start_index += 1
                tokenized_examples["start_positions"].append(token_start_index - 1)
                
                while offsets[token_end_index][1] >= end_char:
                    token_end_index -= 1
                tokenized_examples["end_positions"].append(token_end_index + 1)

    return tokenized_examples

# Apply the mapping 
final_dataset = tokenized_dataset.map(
    add_positions, 
    batched=True,
    remove_columns=["offset_mapping"] if "offset_mapping" in tokenized_dataset.column_names else None
)

print("✅ Success! final_dataset is processed and cleaned.")
print(final_dataset)
```


A random sample was then  selected from the final dataset to verify that the start and end positions correctly map to the answer text.

The tokens corresponding to these positions were decoded back into text and compared.The extracted answer was empty, and the start and end positions pointed to the **[CLS] token**, indicating that the question is unanswerable, confirming  that this  preprocessing step is working correctly, as  it handeled and mapped an unanswerable question as expected.

```python
import random

# 1. Pick a random sample from the final_dataset
index = random.randint(0, len(final_dataset) - 1)
sample = final_dataset[index]

# 2. Get the start and end token positions
start = sample["start_positions"]
end = sample["end_positions"]

# 3. Decode the tokens back to text
input_ids = sample["input_ids"]
# We decode only the tokens between start and end
extracted_answer = tokenizer.decode(input_ids[start:end])

# 4. Get the original text for comparison (referring back to hf_dataset)
# Note: sample_mapping is usually lost after the map, so we check the content
print(f" PRE-PROCESSING VALIDATION (Index {index})")
print(f"✅ Extracted Answer from Tokens: '{extracted_answer}'")

if start == 0 and end == 0:
    print(" This is an 'Unanswerable' question (mapped to [CLS] as expected).")
elif len(extracted_answer.strip()) > 0:
    print("✅ SUCCESS: The token positions correctly align with the answer text.")
else:
    print("❌ WARNING: The positions point to empty or incorrect tokens.")
```

## 4. Model  Training


In this section, the prepared dataset was used to train a question answering model based on **DistilBERT**.The goal is to enable the model to learn how to predict the correct answer span within a given context.

This process includes loading the pre-trained model, splitting the dataset, defining training settings, and running the training process, followed by analysing the model’s performance.

 **Load model**

In this step, the selected question answering model was loaded from the Hugging Face Transformers library.

The **DistilBERT model (`distilbert-base-uncased`)** was chosen for its efficiency,as it retains approximately **97% of BERT’s performance** while being **40% smaller** and about **60% faster** during inference.Hence, given the dataset size, this model was considered a practical choice in terms of both efficiency and training time.

The model is pre-trained on large text corpora and will be fine-tuned on the dataset to learn how to predict the start and end positions of answers in a given context.


```python
from transformers import AutoModelForQuestionAnswering
# Using DistilBERT for speed, 
model = AutoModelForQuestionAnswering.from_pretrained("distilbert-base-uncased")
```

 **Train Validation Test split**

The dataset was divided into training, validation, and test sets using a two-stage splitting approach. 

First, 80% of the data was used for training, while the remaining 20% was set aside as holdout set. This holdout portion was then equally split into validation and test sets.

This resulted in:
- **3,781 training samples**  
- **473 validation samples**  
- **473 test samples**  

A fixed seed (`seed=42`) was used to ensure reproducibility.

The validation set is used to monitor performance during training and guide model selection, while the test set is kept separate and only used at the end to evaluate the model’s generalization performance.

```python
# Split 1: 80% for training, 20% for the holdout set
split_1 = final_dataset.train_test_split(test_size=0.2, seed=42)
train_dataset = split_1["train"]

# Split 2: Divide holdout set into 10% Validation and 10% Test
split_2 = split_1["test"].train_test_split(test_size=0.5, seed=42)
eval_dataset = split_2["train"]
test_dataset = split_2["test"]

# Verify split sizes
print(f"✅ Training samples: {len(train_dataset)}")
print(f"✅ Validation samples: {len(eval_dataset)}")
print(f"✅ Test samples: {len(test_dataset)}")
```

**Training**

In this step, the training process was configured using the Hugging Face `Trainer` API.

Training parameters were defined to control how the model learns, including the learning rate, batch size, and number of epochs. The model was set to train for **3 epochs** with a learning rate of **2e-5**, which is a standard choice for BERT-based models. 

Evaluation and checkpoint saving were performed at the end of each epoch to monitor performance and preserve progress.

An **early stopping mechanism** was also applied, which stops training if the model stops improving on the validation set,to help prevent overfitting and reduce unnecessary training time.

The `Trainer` class was then initialized by combining the model, training arguments, datasets, and tokenizer. 

Finally, the training process was started, allowing the model to learn.

```python
from transformers import TrainingArguments, Trainer, EarlyStoppingCallback

# Training Settings
training_args = TrainingArguments(
    output_dir="./results",          
    evaluation_strategy="epoch",     # Evaluate at the end of each epoch
    save_strategy="epoch",           # Save checkpoint at the end of each epoch
    learning_rate=2e-5,              # Standard BERT learning rate
    per_device_train_batch_size=16,  
    per_device_eval_batch_size=16,    
    num_train_epochs=3,               
    weight_decay=0.01,               
    logging_dir="./logs",            
    logging_steps=50,                
    load_best_model_at_end=True,      # Ensure the best version is kept
    report_to="none"                 
)

# Initialize Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    tokenizer=tokenizer,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=1)]
)

# Launch Training
trainer.train()
```

Training was conducted for **3 epochs** (711 optimizer steps, batch size 16, learning rate 2e-5, weight decay 0.01) with early stopping applied based on validation loss.

The observed losses are summarised below:

| Epoch | Train Loss (avg) | Eval Loss |
|-------|------------------|-----------|
| 1     | ~2.90            | **2.578** |
| 2     | ~2.52            | **2.534** |
| 3     | ~2.38            | **2.546** |

The results show a consistent decrease in training loss, indicating that the model is actively learning. The validation loss reaches its lowest value at **epoch 2**, after which it slightly increases at epoch 3, suggesting the beginning of mild overfitting.

Thanks to the early stopping mechanism, the model automatically retained the best-performing checkpoint from **epoch 2**.

The total training time was approximately **66 minutes**, which aligns well with the expected efficiency of the DistilBERT model.



## 5. Model Evaluation

In this section, the performance of our trained model is evaluated using both quantitative metrics **(Exact Match (EM),F1 Score)** and qualitative analysis. This helps assess how well the model generalizes to unseen data and how accurately it extracts answers from the given context.

After training, the model was evaluated on the held-out test set (**473 samples**) to measure its generalization performance.

The final results are summarised below:

- **Exact Match (EM): 50.11%**  
- **F1 Score (Token-level): 50.79%**  
- **Test Loss:** 2.6318  

The Exact Match (EM) score measures how often the predicted answer exactly matches the ground truth, while the F1 score evaluates the overlap between predicted and true answers, making it more suitable for partial matches.



- Both metrics are clearly above random performance  for an extractive QA task and are within the expected range for a DistilBERT model fine-tuned for only 3 epochs on ~3.8k training pairs of specialised clinical text.

- The small gap between EM and F1 (around 0.7%) indicates that when the model is right it tends to be right , it usually predicts the answer exactly, with few partially correct answers,which is useful in clinical settings where precise answers are preferred.

- From sample outputs, the model has learned how to handle unanswerable questions by predicting the **[CLS] token**. However, some errors occur when the model predicts no answer even when a correct answer exists, meaning it sometimes **over-abstains**.


```python
import numpy as np
import pandas as pd

# 1. GENERATE PREDICTIONS
test_predictions = trainer.predict(test_dataset)
start_logits = test_predictions.predictions[0]
end_logits = test_predictions.predictions[1]
# 2. DECODE TOKENS TO TEXT ANSWERS
final_predicted_answers = []
final_true_answers = []

for i in range(len(test_dataset)):
    start_idx = np.argmax(start_logits[i])
    end_idx = np.argmax(end_logits[i])
    
    # Get the input IDs from the test set
    input_ids = test_dataset[i]["input_ids"]
    
    # Decode Model Prediction: Convert token IDs back to a string
    pred_tokens = tokenizer.convert_ids_to_tokens(input_ids[start_idx:end_idx + 1])
    pred_text = tokenizer.convert_tokens_to_string(pred_tokens).strip()
    final_predicted_answers.append(pred_text)

    # Decode Ground Truth: Convert the actual labels back to a string
    t_start = test_dataset[i]["start_positions"]
    t_end = test_dataset[i]["end_positions"]
    true_tokens = tokenizer.convert_ids_to_tokens(input_ids[t_start:t_end + 1])
    true_text = tokenizer.convert_tokens_to_string(true_tokens).strip()
    final_true_answers.append(true_text)

# 3. DEFINE EVALUATION METRICS 
def normalize_text(text):
    """Lowercases and cleans text to ensure fair comparison"""
    return text.lower().strip()

def compute_f1_metric(prediction, truth):
    """Calculates token-level overlap (Precision/Recall) between model and truth"""
    pred_tokens = normalize_text(prediction).split()
    truth_tokens = normalize_text(truth).split()
    
    if not pred_tokens or not truth_tokens:
        return 1 if pred_tokens == truth_tokens else 0
    
    common = set(pred_tokens) & set(truth_tokens)
    if len(common) == 0:
        return 0
    
    precision = len(common) / len(pred_tokens)
    recall = len(common) / len(truth_tokens)
    return 2 * (precision * recall) / (precision + recall)

# 4. COMPUTE FINAL SCORES
# Exact Match (EM): 1 if exactly the same string, 0 otherwise
em_list = [1 if normalize_text(p) == normalize_text(t) else 0 
           for p, t in zip(final_predicted_answers, final_true_answers)]

# F1 Score: Measures partial overlap (more important for medical QA)
f1_list = [compute_f1_metric(p, t) for p, t in zip(final_predicted_answers, final_true_answers)]

final_em = np.mean(em_list) * 100
final_f1 = np.mean(f1_list) * 100

# 5. PRINT SUMMARY REPORT
print("\n" + "="*50)
print("FINAL MODEL PERFORMANCE REPORT")
print("="*50)
print(f"Total Test Samples:    {len(test_dataset)}")
print(f"Test Loss (Final):     {test_predictions.metrics['test_loss']:.4f}")
print("-" * 50)
print(f"EXACT MATCH (EM):      {final_em:.2f}%")
print(f"F1 SCORE (TOKEN):      {final_f1:.2f}%")
print("="*50)
# Quick verification of the first 3 samples 
print("\n SAMPLE OUTPUTS FOR DISSERTATION DISCUSSION:")
for i in range(3):
    print(f"\nExample {i+1}:")
    print(f"  - Ground Truth: {final_true_answers[i]}")
    print(f"  - Model Answer: {final_predicted_answers[i]}")
    print(f"  - F1 Score:     {f1_list[i]:.4f}")
```

Beyond the numerical metrics, individual test-set predictions were also inspected manually to understand where the model succeeds and where it fails.

The results show that the model is able to capture relevant clinical information from the context, often retrieving meaningful parts of the correct answer. However, in some cases, the predicted answers are overly long or include extra surrounding text, indicating difficulty in precisely identifying the exact answer span.

In addition, there are instances where the model predicts the **[CLS] token** even when a valid answer exists, highlighting a tendency to **over-abstain**. While this behavior is useful for correctly handling unanswerable questions, it can lead to missed answers in certain cases.

## 6. Results Discussion

**Summary of findings**

- A custom SQuAD v2-style dataset of **4,727 QA pairs** (3,676 answerable and 1,051 unanswerable) was successfully constructed, covering key clinical themes related to pressure ulcer management.

- Fine-tuning the **DistilBERT (`distilbert-base-uncased`)** model for 3 epochs achieved a best validation loss of **2.534** at epoch 2, with early stopping correctly selecting this checkpoint.

- On the held-out test set (**473 samples**), the model achieved **Exact Match (EM) = 50.11%** and **F1 = 50.79%**, with a very small gap between the two metrics, indicating that predictions are typically either fully correct or incorrect.

- Qualitative analysis shows that the model can extract clinically meaningful and contextually relevant answers to unseen questions.


### Strengths

1. **Reliable abstention behaviour** – The SQuAD v2 approach was successfully implemented, with the model correctly predicting **[CLS]** for unanswerable questions, a critical feature for safe clinical applications.  

2. **Accurate span extraction** – When the model predicts an answer, it is often exact or closely matches the ground truth, reducing the risk of misleading or hallucinated outputs.  

3. **Reproducible pipeline** – The use of fixed random seeds, consistent dataset splits, and dataset serialization ensures the entire workflow is reproducible and transparent.  

### Limitations

1. **Limited dataset size** – Although sufficient for demonstrating the approach, the relatively small dataset (**4.7k QA pairs**) restricts performance and domain coverage.  

2. **Over-abstention tendency** – The model sometimes predicts **[CLS]** for answerable questions, suggesting a bias toward abstention, likely influenced by the proportion of unanswerable examples.  

3. **Single-run evaluation** – Results are based on one training run, without measuring variability across multiple seeds or runs.  

4. **Model capacity constraints** – While efficient, DistilBERT has lower representational capacity compared to larger or domain-specific models, which may limit performance in complex clinical contexts.  


### Future Work

- Fine-tune domain-specific models such as **BioBERT, ClinicalBERT, or PubMedBERT** to improve understanding of medical terminology.  
- Adjust the balance between answerable and unanswerable questions to reduce over-abstention.  
- Incorporate additional evaluation metrics such as **BERTScore** to better capture semantic similarity.  
- Perform cross-validation or multi-seed experiments to obtain more robust performance estimates.
